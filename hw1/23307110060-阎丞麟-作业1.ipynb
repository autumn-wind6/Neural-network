### 2
Write a Python function that takes a list and returns a new list with unique elements of the first list

```python
def unique_list(list):
    unique = set()
    new_list = []
    for i in list:
        if i not in unique:
            unique.add(i)
            new_list.append(i)
    return new_list

input_list = [1, 2, 3, 3, 3, 3, 4, 5]
unique_list(input_list)
```

Write a Python function that checks whether a passed string is palindrome or not.   
A palindrome is a word,phrase, or sequence that reads the same backward as forward.  
For example, both “madam” and “nurses run”are palindromes

```python
def palindrome(string):
    cleaned = ''.join(string.lower().split())
    return cleaned == cleaned[::-1]
    
print(palindrome("madam"))        # True
print(palindrome("nurses run"))   # True
print(palindrome("hello"))
```

### 5 
Write a Python program to add two binary numbers

```python
def add_binary(a, b):
    result = []
    carry = 0
    i, j = len(a) - 1, len(b) - 1

    while i >= 0 or j >= 0 or carry:
        total = carry
        if i >= 0:
            total += int(a[i])
            i -= 1
        if j >= 0:
            total += int(b[j])
            j -= 1
        result.append(str(total % 2))
        carry = total // 2

    return ''.join(reversed(result))

print(add_binary('11', '1')) 
```

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def addTwoNumbers(l1: ListNode, l2: ListNode) -> ListNode:
    dummy = ListNode(0)
    curr = dummy
    carry = 0

    while l1 or l2 or carry:
        v1 = l1.val if l1 else 0
        v2 = l2.val if l2 else 0
        total = v1 + v2 + carry
        carry = total // 10
        curr.next = ListNode(total % 10)
        curr = curr.next

        if l1: l1 = l1.next
        if l2: l2 = l2.next

    return dummy.next


l1 = ListNode(2)
l1.next = ListNode(4)
l1.next.next = ListNode(3)
l2 = ListNode(5)
l2.next = ListNode(6)
l2.next.next = ListNode(4)

def print_linked_list(node):
    result = []
    while node:
        result.append(str(node.val))
        node = node.next
    print(" -> ".join(result))


result = addTwoNumbers(l1, l2)
print_linked_list(result)
```

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr


print(bubble_sort([64, 34, 25, 12, 22, 11, 90]))
```

```python
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    
    result.extend(left[i:])
    result.extend(right[j:])
    return result

print(merge_sort([38, 27, 43, 3, 9, 82, 10]))
```

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    
    pivot = arr[len(arr) // 2] 
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    
    return quick_sort(left) + middle + quick_sort(right)

print(quick_sort([64, 34, 25, 12, 22, 11, 90]))
```

```python
def shell_sort(arr):
    n = len(arr)
    gap = n // 2        
    
    while gap > 0:
        for i in range(gap, n):
            temp = arr[i]
            j = i
            while j >= gap and arr[j - gap] > temp:
                arr[j] = arr[j - gap]
                j -= gap
            arr[j] = temp
        gap //= 2         
    return arr

print(shell_sort([12, 34, 54, 2, 3, 1, 99, 23]))
```

```python
import torch

# 1. 生成线性回归数据 (y 是连续值，不再是 0 或 1)
torch.manual_seed(0)
X = torch.randn(100, 2)
true_w = torch.tensor([[2.0], [-3.0]])
true_b = torch.tensor([0.5])

# 线性回归公式: y = Xw + b + 噪声
y = X @ true_w + true_b + torch.randn(100, 1) * 0.01 

# 2. 初始化参数
w = torch.randn(2, 1, requires_grad=True)
b = torch.zeros(1, requires_grad=True)

# 3. 训练
lr = 0.1
for epoch in range(100):
    # forward: 线性预测，不需要 sigmoid
    y_pred = X @ w + b
    
    # loss: 使用均方误差 (MSE)
    loss = torch.mean((y_pred - y)**2)

    # backward
    loss.backward()

    # 更新参数
    with torch.no_grad():
        w -= lr * w.grad
        b -= lr * b.grad

        # 清空梯度
        w.grad.zero_()
        b.grad.zero_()

print("训练出的 w:", w.detach().flatten())
print("真实的 w:  ", true_w.flatten())
```

```python
import torch

# 1. 生成二分类数据
torch.manual_seed(0)
X = torch.randn(100, 2)

true_w = torch.tensor([[2.0], [-3.0]])
true_b = torch.tensor([0.5])

logits = X @ true_w + true_b
y = (torch.sigmoid(logits) > 0.5).float()

# 2. 初始化参数
w = torch.randn(2, 1, requires_grad=True)
b = torch.zeros(1, requires_grad=True)

# 3. 训练
lr = 0.1
for epoch in range(200):

    # forward
    logits = X @ w + b
    y_pred = torch.sigmoid(logits)

    # BCE loss
    loss = -(y * torch.log(y_pred) + 
             (1 - y) * torch.log(1 - y_pred)).mean()

    # backward
    loss.backward()

    # 更新参数
    with torch.no_grad():
        w -= lr * w.grad
        b -= lr * b.grad

        w.grad.zero_()
        b.grad.zero_()


print("w:", w)
print("b:", b)
```

### 14
Add a Frobenius norm penalty for the weight $w$ in your SVM model by two different ways: (1) use a PyTorch function to calculate the norm; (2) implement the code by yourself.

Hint: $\|A\|_F = \left(\sum_{i,j} |a_{ij}|^2\right)^{1/2}$. 训练时常用 $\|w\|_F^2=\sum_{i,j}w_{ij}^2$ 作为光滑的 L2 正则项。

```python
import torch

torch.manual_seed(0)

X = torch.randn(100, 2)
true_w = torch.tensor([[2.0], [-3.0]])
true_b = torch.tensor([0.5])

y = X @ true_w + true_b
y = torch.sign(y)

lam = 0.1  # Frobenius / L2 正则强度

# ---------- (1) 用 PyTorch 计算 Frobenius 范数，再加惩罚项 ----------
torch.manual_seed(1)  # 与 (2) 同种子初始化 w，否则 X 已消耗 RNG，两次 randn(w) 不同
w = torch.randn(2, 1, requires_grad=True)
b = torch.zeros(1, requires_grad=True)
lr = 0.01
for _ in range(200):
    scores = X @ w + b
    hinge = torch.clamp(1 - y * scores, min=0).mean()
    # ||w||_F^2 = torch.norm(w, p='fro')**2
    loss = hinge + lam * torch.norm(w, p="fro") ** 2
    loss.backward()
    with torch.no_grad():
        w -= lr * w.grad
        b -= lr * b.grad
        w.grad.zero_()
        b.grad.zero_()
print("(1) torch.norm  w =", w.squeeze().tolist(), "  b =", b.item())

# ---------- (2) 按定义手写：||A||_F = sqrt( sum_{i,j} |a_ij|^2 )，惩罚用 ||w||_F^2 ----------
torch.manual_seed(1)  # 与 (1) 相同初值
w = torch.randn(2, 1, requires_grad=True)
b = torch.zeros(1, requires_grad=True)
for _ in range(200):
    scores = X @ w + b
    hinge = torch.clamp(1 - y * scores, min=0).mean()
    fro_sq = torch.sum(w ** 2)  # sum |w_ij|^2 = ||w||_F^2（与先 sqrt 再平方相同）
    loss = hinge + lam * fro_sq
    loss.backward()
    with torch.no_grad():
        w -= lr * w.grad
        b -= lr * b.grad
        w.grad.zero_()
        b.grad.zero_()
print("(2) 手写 sum w_ij^2  w =", w.squeeze().tolist(), "  b =", b.item())
```

Write a custom **dataset class** for loading CIFAR-10 compatible with PyTorch `DataLoader`.

1. Inherit from PyTorch `Dataset`.  
2. Load the image file and save it in a proper way.  
3. Override `__getitem__` and `__len__`.

```python
import os
import pickle
import numpy as np
from PIL import Image
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

class CIFAR10PickleDataset(Dataset):
    """从官方 pickle 批次文件读取 CIFAR-10，继承 Dataset。"""

    train_list = [f"data_batch_{i}" for i in range(1, 6)]
    test_list = ["test_batch"]

    def __init__(self, root, train=True, transform=None):
        self.root = root
        self.train = train
        self.transform = transform
        self.data = []
        self.targets = []

        base_folder = os.path.join(root, "cifar-10-batches-py")
        file_list = self.train_list if train else self.test_list
        for file_name in file_list:
            path = os.path.join(base_folder, file_name)
            with open(path, "rb") as f:
                entry = pickle.load(f, encoding="latin1")
            self.data.append(entry["data"])
            if train:
                self.targets.extend(entry["labels"])
            else:
                self.targets = entry["labels"]
        self.data = np.vstack(self.data)
        # 每行 3072 = 3×32×32，按 R/G/B 通道展开 → reshape 为 N×3×32×32
        self.data = self.data.reshape(-1, 3, 32, 32)
        self.data = self.data.transpose((0, 2, 3, 1))  # NHWC，uint8

    def __len__(self):
        return len(self.targets)

    def __getitem__(self, index):
        img = self.data[index]
        img = Image.fromarray(img.astype(np.uint8))
        target = self.targets[index]
        if self.transform is not None:
            img = self.transform(img)
        return img, target


# 需先运行「下载 CIFAR-10」那一格
custom_train = CIFAR10PickleDataset(CIFAR_ROOT, train=True, transform=None)
print("自定义 Dataset 训练集长度:", len(custom_train))
img0, y0 = custom_train[0]
print("第 0 个样本: PIL 图尺寸", img0.size, "标签", y0)

custom_train_tensor = CIFAR10PickleDataset(
    CIFAR_ROOT, train=True, transform=transforms.ToTensor()
)
loader_smoke = DataLoader(custom_train_tensor, batch_size=16, shuffle=False)
xb, yb = next(iter(loader_smoke))
print("DataLoader 一批张量形状:", xb.shape, "标签:", yb[:4].tolist(), "...")
```

### 19
Performance testing for data loading: run **one epoch** of iterating the CIFAR-10 training set with `DataLoader`, and record loading time for:

- **batch_size:** 1, 4, 64, 1024  
- **num_workers:** 0, 1, 4, 16  
- **pin_memory:** `True` and `False`

```python
import time
import torch
from torch.utils.data import DataLoader
from torchvision import transforms, datasets

# 1. 准备数据集（与前面 Q16 一致，用 torchvision 自带实现；数据已在 ./data 则不会重复下载）
_ds = datasets.CIFAR10(
    root=CIFAR_ROOT,
    train=True,
    download=True,
    transform=transforms.ToTensor(),
)

batch_sizes = [1, 4, 64, 1024]
num_workers_list = [0, 1, 4, 16]
pin_list = [False, True]

# 预热函数：消除首次加载的系统延迟（I/O缓存、库初始化等）
def warm_up(dataset):
    temp_loader = DataLoader(dataset, batch_size=64, num_workers=0)
    for i, _ in enumerate(temp_loader):
        if i > 5: break 

warm_up(_ds)

rows = []
print(f"{'Batch':>6} | {'Workers':>8} | {'Pin':>6} | {'Time (s)':>10} | {'Img/s':>10}")
print("-" * 55)
for bs in batch_sizes:
    for nw in num_workers_list:
        for pin in pin_list:
            loader = DataLoader(
                _ds,
                batch_size=bs,
                shuffle=True,
                num_workers=nw,
                pin_memory=pin,
                drop_last=False
            )
            
            # 确保 GPU 任务完全停止后再开始计时
            if torch.cuda.is_available():
                torch.cuda.synchronize()
            
            t0 = time.perf_counter()
            
            # 开始跑一轮 epoch
            for _x, _y in loader:
                pass
            
            # 确保所有异步的数据拷贝/处理彻底完成
            if torch.cuda.is_available():
                torch.cuda.synchronize()
                
            elapsed = time.perf_counter() - t0
            
            # 计算每秒处理的图像数 (Throughput)
            imgs_per_sec = len(_ds) / elapsed
            rows.append((bs, nw, pin, elapsed, imgs_per_sec))
            print(
                f"{bs:>6} | {nw:>8} | {str(pin):>6} | {elapsed:>10.3f} | {imgs_per_sec:>10.1f}",
                flush=True,
            )

fastest = min(rows, key=lambda x: x[3])
print(f"\n最佳配置: Batch={fastest[0]}, Workers={fastest[1]}, Pin={fastest[2]} (用时: {fastest[3]:.3f}s)")
```

### 20
Calculate the **mean** and **standard deviation** of the CIFAR-10 **training set** for each of the three RGB channels.

```python
import torch
from torch.utils.data import DataLoader  
from torchvision import transforms

# 使用 [0,1] 张量统计（ToTensor），与常见预处理一致
_train_stats = CIFAR10PickleDataset(
    CIFAR_ROOT, train=True, transform=transforms.ToTensor()
)
loader_stats = DataLoader(_train_stats, batch_size=1024, shuffle=False, num_workers=0)

n_pixels = 0
channel_sum = torch.zeros(3)
channel_sq_sum = torch.zeros(3)

for imgs, _ in loader_stats:
    # imgs: (B, 3, 32, 32), 值域 [0, 1]
    b, c, h, w = imgs.shape
    flat = imgs.view(b, c, -1)
    channel_sum += flat.sum(dim=(0, 2))
    channel_sq_sum += (flat ** 2).sum(dim=(0, 2))
    n_pixels += b * h * w

mean = channel_sum / n_pixels
std = (channel_sq_sum / n_pixels - mean ** 2).clamp_min(0).sqrt()

print("CIFAR-10 训练集各通道 mean:", [round(x, 4) for x in mean.tolist()])
print("CIFAR-10 训练集各通道 std :", [round(x, 4) for x in std.tolist()])
```

### 21
Image to character painting（RGB → 灰度 → 字符 → 保存 `.txt`）

```python
import numpy as np
from pathlib import Path
from PIL import Image

# 字符表：由暗到亮（可自定义）；灰度 0→最左、1→最右
CHARS = "@%#*+=-:. "

def rgb_to_gray(rgb_arr):
    # PDF：gray = 0.2126*r + 0.7152*g + 0.0722*b；RGB 先归一化到 [0,1] 再代入
    r = rgb_arr[..., 0].astype(np.float64) / 255.0
    g = rgb_arr[..., 1].astype(np.float64) / 255.0
    b = rgb_arr[..., 2].astype(np.float64) / 255.0
    return 0.2126 * r + 0.7152 * g + 0.0722 * b


def image_to_char_painting(pil_img, out_width=80):
    h0, w0 = pil_img.size[1], pil_img.size[0]
    out_height = max(1, int(h0 / w0 * out_width * 0.45))
    try:
        resample = Image.Resampling.LANCZOS
    except AttributeError:
        resample = Image.LANCZOS
    small = np.asarray(pil_img.convert("RGB").resize((out_width, out_height), resample))
    gray = rgb_to_gray(small)
    # 按灰度直接分档（不做 min-max 拉伸，避免与 PDF 要求的灰度对应关系不一致）
    idx = np.clip(
        np.round(gray * (len(CHARS) - 1)).astype(np.int32), 0, len(CHARS) - 1
    )
    lines = []
    for row in idx:
        lines.append("".join(CHARS[i] for i in row))
    return "\n".join(lines)


# 使用 sheep 图片：放在与本 ipynb 同一目录，命名为 sheep.jpg / sheep.png 等
# 若文件名不同，可改为：pil_img = Image.open(r"绝对路径\你的图.png").convert("RGB")
_img_exts = {".jpg", ".jpeg", ".png", ".webp", ".bmp", ".gif"}
_sheep = [p for p in Path(".").glob("sheep.*") if p.suffix.lower() in _img_exts]
if not _sheep:
    raise FileNotFoundError("未找到 sheep 图片，请将 sheep.jpg 或 sheep.png 放在当前工作目录")
pil_img = Image.open(sorted(_sheep)[0]).convert("RGB")

text_art = image_to_char_painting(pil_img, out_width=72)
out_path = "char_painting_output.txt"
with open(out_path, "w", encoding="utf-8") as f:
    f.write(text_art)
lines = text_art.splitlines()
print("已写入", out_path)
print("行数:", len(lines), "列宽:", len(lines[0]) if lines else 0)
print("\n--- 字符画全文 ---\n")
print(text_art)
```

### 23
Bilinear interpolation（双线性插值），坐标为 **1-indexed**，与 PDF 测试一致。

```python
import numpy as np


def BilinearInterpolation(A, xy):
    """A 为二维数组；xy=(x,y) 为 1-indexed 连续坐标。"""
    A = np.asarray(A, dtype=float)
    nrow, ncol = A.shape
    x, y = float(xy[0]), float(xy[1])
    if 1 <= x <= nrow and 1 <= y <= ncol:
        if abs(x - round(x)) < 1e-12 and abs(y - round(y)) < 1e-12:
            return float(A[int(round(x)) - 1, int(round(y)) - 1])
    x1, x2 = int(np.floor(x)), int(np.ceil(x))
    y1, y2 = int(np.floor(y)), int(np.ceil(y))
    x1 = max(1, min(x1, nrow))
    x2 = max(1, min(x2, nrow))
    y1 = max(1, min(y1, ncol))
    y2 = max(1, min(y2, ncol))
    if x1 == x2 and y1 == y2:
        return float(A[x1 - 1, y1 - 1])
    if x1 == x2:
        x2 = min(x1 + 1, nrow)
    if y1 == y2:
        y2 = min(y1 + 1, ncol)
    f11 = A[x1 - 1, y1 - 1]
    f12 = A[x1 - 1, y2 - 1]
    f21 = A[x2 - 1, y1 - 1]
    f22 = A[x2 - 1, y2 - 1]
    tx = 0.0 if x2 == x1 else (x - x1) / (x2 - x1)
    ty = 0.0 if y2 == y1 else (y - y1) / (y2 - y1)
    return float(
        (1 - tx) * (1 - ty) * f11
        + (1 - tx) * ty * f12
        + tx * (1 - ty) * f21
        + tx * ty * f22
    )


A = ((110, 120, 130), (210, 220, 230), (310, 320, 330))
print(BilinearInterpolation(A, (1, 1)) )
print(abs(BilinearInterpolation(A, (2.5, 2.5)) ) )
```

### 25
从数组中截取固定形状子块，中心对齐给定元素，越界用 `fill` 填充。

```python
import numpy as np


def extract_subpart(Z, shape, fill, position):
    Z = np.asarray(Z)
    h, w = shape
    cy, cx = position
    top = cy - h // 2
    left = cx - w // 2
    out = np.full(shape, fill, dtype=Z.dtype)
    for i in range(h):
        for j in range(w):
            zi, zj = top + i, left + j
            if 0 <= zi < Z.shape[0] and 0 <= zj < Z.shape[1]:
                out[i, j] = Z[zi, zj]
            else:
                out[i, j] = fill
    return out


Z = np.array(
    [
        [3, 6, 8, 5, 9],
        [4, 9, 0, 0, 9],
        [6, 1, 4, 0, 8],
        [9, 1, 2, 0, 9],
        [4, 1, 7, 5, 0],
    ]
)
print("Z=\n", Z)
print("截取结果:\n", extract_subpart(Z, (4, 4), 0, (1, 1)))
```

### 26
仅用 Python 列表实现二维矩阵运算（不用 NumPy）。

```python
def _shape(A):
    return len(A), len(A[0]) if A else 0


def add(A, B):
    return [[A[i][j] + B[i][j] for j in range(len(A[0]))] for i in range(len(A))]


def subtract(A, B):
    return [[A[i][j] - B[i][j] for j in range(len(A[0]))] for i in range(len(A))]


def scalar_multiply(A, k):
    return [[A[i][j] * k for j in range(len(A[0]))] for i in range(len(A))]


def multiply(A, B):
    rA, cA = _shape(A)
    rB, cB = _shape(B)
    assert cA == rB
    return [
        [sum(A[i][k] * B[k][j] for k in range(cA)) for j in range(cB)] for i in range(rA)
    ]


def identity(n):
    return [[1 if i == j else 0 for j in range(n)] for i in range(n)]


def transpose(A):
    r, c = _shape(A)
    return [[A[j][i] for j in range(r)] for i in range(c)]


def inverse(A):
    n = len(A)
    aug = [row[:] + [float(i == j) for j in range(n)] for i, row in enumerate(A)]
    for col in range(n):
        piv = col
        for r in range(col + 1, n):
            if abs(aug[r][col]) > abs(aug[piv][col]):
                piv = r
        aug[col], aug[piv] = aug[piv], aug[col]
        if abs(aug[col][col]) < 1e-12:
            raise ValueError("不可逆")
        div = aug[col][col]
        aug[col] = [x / div for x in aug[col]]
        for r in range(n):
            if r != col and abs(aug[r][col]) > 1e-12:
                f = aug[r][col]
                aug[r] = [aug[r][c] - f * aug[col][c] for c in range(2 * n)]
    return [row[n:] for row in aug]


matrix_a = [[12, 10], [3, 9]]
matrix_b = [[3, 4], [7, 4]]
matrix_c = [
    [11, 12, 13, 14],
    [21, 22, 23, 24],
    [31, 32, 33, 34],
    [41, 42, 43, 44],
]
matrix_d = [[3, 0, 2], [2, 0, -2], [0, 1, 1]]

print("add:", add(matrix_a, matrix_b))
print("subtract:", subtract(matrix_a, matrix_b))
print("scalar_multiply:", scalar_multiply(matrix_b, 3))
print("multiply:", multiply(matrix_a, matrix_b))
print("identity(3):", identity(3))
print("transpose:", transpose(matrix_c))
inv_d = inverse(matrix_d)
print("inverse(matrix_d):", inv_d)
```

### 27
最大公约数 GCD（欧几里得算法）。

```python
def gcd(a, b):
    a, b = abs(int(a)), abs(int(b))
    if a == 0:
        return b
    if b == 0:
        return a
    while b:
        a, b = b, a % b
    return a


print(gcd(3, 5), gcd(6, 3), gcd(-2, 6), gcd(0, 3))
```

```python
def tiling_ways(n):
    if n < 0:
        return 0
    f = [0] * (n + 1)
    f[0] = 1
    if n >= 1:
        f[1] = 1
    for i in range(2, n + 1):
        f[i] = f[i - 1] + 2 * f[i - 2]
    return f[n]


def tiling_ways_bruteforce(n):
    """DFS 枚举，用于验证小 n。"""
    H, W = 2, n
    g = [[False] * W for _ in range(H)]

    def dfs():
        fi = fj = None
        for i in range(H):
            for j in range(W):
                if not g[i][j]:
                    fi, fj = i, j
                    break
            if fi is not None:
                break
        if fi is None:
            return 1
        i, j = fi, fj
        cnt = 0
        if i == 0 and not g[1][j]:
            g[0][j] = g[1][j] = True
            cnt += dfs()
            g[0][j] = g[1][j] = False
        if j + 1 < W and i == 0 and not g[0][j + 1]:
            g[0][j] = g[0][j + 1] = True
            cnt += dfs()
            g[0][j] = g[0][j + 1] = False
        if j + 1 < W and i == 1 and not g[1][j + 1]:
            g[1][j] = g[1][j + 1] = True
            cnt += dfs()
            g[1][j] = g[1][j + 1] = False
        if i == 0 and j + 1 < W and not g[0][j + 1] and not g[1][j] and not g[1][j + 1]:
            g[0][j] = g[0][j + 1] = g[1][j] = g[1][j + 1] = True
            cnt += dfs()
            g[0][j] = g[0][j + 1] = g[1][j] = g[1][j + 1] = False
        return cnt

    return dfs()


for n in range(1, 9):
    a, b = tiling_ways(n), tiling_ways_bruteforce(n)
    print(f"n={n}: 递推={a}, DFS={b}, 一致={a == b}")
```

### 29
密码合法性检验（长度 6–12；含小写、大写、数字、`$#@` 各至少一个），多密码逗号分隔输入。

```python
import re


def is_valid_password(p: str) -> bool:
    p = p.strip()
    if not (6 <= len(p) <= 12):
        return False
    if not re.search(r"[a-z]", p):
        return False
    if not re.search(r"[A-Z]", p):
        return False
    if not re.search(r"[0-9]", p):
        return False
    if not re.search(r"[$#@]", p):
        return False
    return True


def filter_passwords(line: str):
    parts = [x.strip() for x in line.split(",")]
    return [p for p in parts if is_valid_password(p)]


example = "ABd1234@1,a F1#,2w3E*,2We3345"
valid = filter_passwords(example)
print("合法密码:", ",".join(valid))
```
